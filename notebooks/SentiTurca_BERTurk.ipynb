```python
import subprocess
from pathlib import Path

REPOSITORY = Path('/kaggle/working/Sagduyu')
if not REPOSITORY.exists():
    subprocess.run(['git', 'clone', 'https://github.com/FARADTechnologies/Sagduyu.git', str(REPOSITORY)], check=True)
OUTPUT = Path('/kaggle/working/sentiturca_results.json')
CHECKPOINTS = Path('/kaggle/working/sentiturca-checkpoints')
```

```python
subprocess.run([
    'python', str(REPOSITORY / 'experiments/sentiturca_berturk.py'),
    '--output', str(OUTPUT),
    '--checkpoint-dir', str(CHECKPOINTS),
    '--modes', 'raw,normalized',
    '--seeds', '42',
    '--epochs', '3',
], cwd=REPOSITORY, check=True)
```

```python
import json
result = json.loads(OUTPUT.read_text())
print(json.dumps(result['mean'], indent=2))
print('Ürün kabul kapısı:', result['acceptance_gate']['accepted'])
```
